# charger le fichier excel 

```python
from pathlib import Path
import pandas as pd

# Chemin absolu vers le dossier raw
RAW_FOLDER = Path.cwd().parent / "data" / "raw"

# Recherche automatique des fichiers Excel
fichiers_excel = list(RAW_FOLDER.glob("*.xlsx"))

print("Dossier recherché :", RAW_FOLDER)
print("Fichiers trouvés :", fichiers_excel)
```

```python
if not fichiers_excel:
    raise FileNotFoundError("Aucun fichier Excel trouvé dans data/raw")

DATA_PATH = fichiers_excel[0]

print("Fichier utilisé :", DATA_PATH)
print("Existe réellement :", DATA_PATH.exists())

df = pd.read_excel(DATA_PATH, engine="openpyxl")

df.head()
```

```python
df = pd.read_excel(
    DATA_PATH,
    header=1,
    engine="openpyxl"
)

df.head()
```

# Verifions la taille 

```python
print(f"Nombre de lignes : {df.shape[0]}")
print(f"Nombre de colonnes : {df.shape[1]}")
```

# Recherche de Valeurs manquantes : 

```python
valeurs_manquantes = df.isna().sum()

valeurs_manquantes
```

# Verifions les doublons 

```python
nombre_doublons = df.duplicated().sum()

print("Nombre de lignes dupliquées :", nombre_doublons)
```

# Examinons la variable principale 

```python
df["Statut du client"].value_counts(dropna=False)
```

```python
df["Statut du client"].value_counts(
    normalize=True,
    dropna=False
).mul(100).round(2)
```

# Statistiques generales 

```python
df.describe(include="all").T
```

# Examinons les catégories de chaque colonne 

```python
colonnes_categorielles = df.select_dtypes(include="object").columns

for colonne in colonnes_categorielles:
    print(f"\n--- {colonne} ---")
    print(df[colonne].value_counts(dropna=False))
```

# Verifions les colonnes minimales et maximales 

```python
colonnes_numeriques = (
    df.select_dtypes(include="number")
    .columns
    .drop("N° du client")
)

df[colonnes_numeriques].agg(
    ["min", "max", "mean", "median"]
).T
```

# Verifions les espaces cachées 

```python
for colonne in colonnes_categorielles:
    valeurs_avec_espaces = (
        df[colonne]
        .dropna()
        .astype(str)
        .loc[lambda x: x != x.str.strip()]
    )

    print(
        f"{colonne} : "
        f"{len(valeurs_avec_espaces)} valeur(s) avec espaces inutiles"
    )
```

# Creer un resumé automatique de la qualité 

```python
resume_qualite = pd.DataFrame({
    "type": df.dtypes.astype(str),
    "valeurs_manquantes": df.isna().sum(),
    "pourcentage_manquant": df.isna().mean().mul(100).round(2),
    "valeurs_uniques": df.nunique(dropna=False)
})

resume_qualite
```
